### Plots

```python
def plot_correlations(causal_df, outcome, correlation, method_name):
    # Transforming the DataFrame using melt
    causal_df = causal_df[causal_df['method_name'] == method_name]
    correlation_plot_df = causal_df.melt(id_vars=['rule_number'], value_vars=[correlation, 'estimated_effect'],  var_name='type', value_name='value')

    # Create the plot
    plt.figure(figsize=(20, 6))
    barplot = sns.barplot(
        palette='mako',
        data=correlation_plot_df,
        x="rule_number",
        y="value",
        hue="type",
        alpha=0.8, 
        errorbar=None)

    # Improve x-axis readability
    plt.xlabel("Rule Number", fontsize=12)
    plt.ylabel("Value", fontsize=12)
    plt.title(f"{correlation} vs ATE. Outcome [{outcome}], Model [{params['model']}]", fontsize=14)

    # Move the legend outside for clarity
    plt.legend(title="Metrics", bbox_to_anchor=(1.05, 1), loc='upper left')

    # Add vertical grid lines to separate rule numbers
    for rule in correlation_plot_df["rule_number"].unique():
        plt.axvline(x=(rule if isinstance(rule, int) else 24)- 1, color='gray', linestyle='--', alpha=0.2)
    
    output_dir = f"{params['output_plots']}/{params['model']}/{method_name}"
    create_folder(output_dir)
    plt.savefig(f"{output_dir}/{outcome}_{correlation}.png", bbox_inches="tight", dpi=300)
    plt.close()

```
